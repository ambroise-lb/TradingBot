```python
import os
import numpy as np
import pandas as pd
import yfinance as yf

def charger_donnees(ticker, start="2000-01-01", dossier="data"):
    os.makedirs(dossier, exist_ok=True)
    chemin = f"{dossier}/{ticker}_daily.csv"

    if os.path.exists(chemin):
        df = pd.read_csv(chemin, index_col=0, parse_dates=True)
    else:
        df = yf.download(ticker, start=start, auto_adjust=True, multi_level_index=False)
        df.to_csv(chemin)

    df = df.sort_index()
    df = df[~df.index.duplicated(keep="first")]
    df = df.dropna(subset=["Close"])
    df = df[df["Close"] > 0]
    return df


# ---------- 2) FEATURES (uniquement du passe, connu a la cloture de J) ----------

def calcul_rsi(close, periode=14):
    delta = close.diff()
    gain = delta.clip(lower=0)
    perte = -delta.clip(upper=0)
    moy_gain = gain.ewm(alpha=1 / periode, adjust=False).mean()
    moy_perte = perte.ewm(alpha=1 / periode, adjust=False).mean()
    rs = moy_gain / moy_perte
    return 100 - 100 / (1 + rs)


def ajouter_features(df):
    d = df.copy()
    c, v = d["Close"], d["Volume"]
    rend = c.pct_change()

    # ecart relatif du prix a chaque moyenne (comparable dans le temps)
    for n in [50, 200]:
        d[f"d_sma{n}"] = c / c.rolling(n).mean() - 1
        d[f"d_ema{n}"] = c / c.ewm(span=n, adjust=False).mean() - 1
        vwma = (c * v).rolling(n).sum() / v.rolling(n).sum()
        d[f"d_vwma{n}"] = c / vwma - 1

    # tendance : MA50 au-dessus ou en dessous de MA200
    d["sma50_sur_sma200"] = c.rolling(50).mean() / c.rolling(200).mean() - 1

    # momentum (rendements passes)
    for n in [1, 5, 20, 60]:
        d[f"mom{n}"] = c.pct_change(n)

    # volatilite
    d["vol20"] = rend.rolling(20).std()
    d["vol60"] = rend.rolling(60).std()
    d["ratio_vol"] = d["vol20"] / d["vol60"]         # volatilite recente vs normale

    # RSI
    d["rsi14"] = calcul_rsi(c, 14)

    # volume relatif : volume du jour vs moyenne des 20 derniers jours
    d["vol_relatif"] = v / v.rolling(20).mean()

    # pression acheteuse approximative (position de la cloture dans la bougie)
    ampleur = (d["High"] - d["Low"]).replace(0, np.nan)
    d["clv"] = ((c - d["Low"]) - (d["High"] - c)) / ampleur

    # distance au plus haut des 252 derniers jours (drawdown courant)
    d["dd252"] = c / c.rolling(252).max() - 1

    return d


FEATURES = [
    "d_sma50", "d_ema50", "d_vwma50", "d_sma200", "d_ema200", "d_vwma200",
    "sma50_sur_sma200", "mom1", "mom5", "mom20", "mom60",
    "vol20", "vol60", "ratio_vol", "rsi14", "vol_relatif", "clv", "dd252",
]


# ---------- 3) LABEL (le seul endroit ou l'on regarde le futur) ----------

def ajouter_label(d, h=5):
    d = d.copy()
    d["rend_futur"] = d["Close"].shift(-h) / d["Close"] - 1
    d["label"] = (d["rend_futur"] > 0).astype(int)
    d.loc[d["rend_futur"].isna(), "label"] = np.nan   # les h derniers jours n'ont pas de reponse
    return d


# ---------- UTILISATION ----------

H = 5
df = charger_donnees("SPY")
table = ajouter_features(df)
table = ajouter_label(table, h=H)

# pour entrainer : on garde seulement les lignes completes
table_ml = table.dropna(subset=FEATURES + ["rend_futur"]).copy()
table_ml["label"] = table_ml["label"].astype(int)

print(table_ml[FEATURES + ["label"]].tail())
print("Lignes :", len(table_ml))
print("Part de label=1 :", round(table_ml["label"].mean(), 3))
```

```python
df2 = df.copy()
df2.iloc[-1, df2.columns.get_loc("Close")] *= 1.5
f1 = ajouter_features(df)[FEATURES].iloc[:-1]
f2 = ajouter_features(df2)[FEATURES].iloc[:-1]
print("Pas de fuite :", np.allclose(f1.fillna(0), f2.fillna(0)))
```

```python
import lightgbm as lgb
from sklearn.metrics import roc_auc_score

H = 5
SEUIL = 0.55      # a fixer AVANT de regarder les resultats
FRAIS = 0.001

def metriques(rend, nom):
    capital = (1 + rend).cumprod()
    annuel = capital.iloc[-1] ** (252 / len(rend)) - 1
    vol = rend.std() * np.sqrt(252)
    sharpe = rend.mean() / rend.std() * np.sqrt(252)
    drawdown = (capital / capital.cummax() - 1).min()
    print(f"{nom:14s} rendement annuel {annuel:6.1%} | vol {vol:5.1%} | "
          f"Sharpe {sharpe:4.2f} | drawdown max {drawdown:6.1%}")

def walk_forward(table, h, taille_train=252*5, taille_test=252):
    debut = 0
    while debut + taille_train + h < len(table):
        train = table.iloc[debut : debut + taille_train]
        test = table.iloc[debut + taille_train + h : debut + taille_train + h + taille_test]
        yield train, test
        debut += taille_test

# ---- entrainement : un modele par fenetre, teste sur la periode suivante ----
preds = []
for train, test in walk_forward(table_ml, h=H):
    modele = lgb.LGBMClassifier(
        n_estimators=200, learning_rate=0.03, max_depth=3, num_leaves=8,
        min_child_samples=50, subsample=0.8, subsample_freq=1,
        colsample_bytree=0.8, random_state=0, verbose=-1,
    )
    modele.fit(train[FEATURES], train["label"])
    p = modele.predict_proba(test[FEATURES])[:, 1]      # probabilite de hausse
    preds.append(pd.Series(p, index=test.index))

proba = pd.concat(preds)
res = table_ml.loc[proba.index].copy()
res["proba"] = proba
```

```python
import matplotlib.pyplot as plt

print("Periode testee :", res.index[0].date(), "->", res.index[-1].date())
print("AUC :", round(roc_auc_score(res["label"], res["proba"]), 3), "(0.5 = hasard)")
print("Baseline 'toujours hausse' :", round(res["label"].mean(), 3))
print("Precision du modele (seuil 0.5) :", round(((res["proba"] > 0.5) == res["label"]).mean(), 3))

# ---- backtest : position de demain = decision d'aujourd'hui ----
rend = res["Close"].pct_change()

def strategie(signal):
    pos = signal.astype(int).shift(1)
    trades = pos.diff().abs()
    return (pos * rend - trades * FRAIS).dropna(), int(trades.sum())

rend_ia, nb_ia = strategie(res["proba"] > SEUIL)
rend_sma, nb_sma = strategie(res["d_sma200"] > 0)       # = Close > SMA200
rend_bh = rend.dropna()

metriques(rend_bh, "Buy and hold")
metriques(rend_sma, "SMA200")
print("  trades :", nb_sma)
metriques(rend_ia, "IA (LightGBM)")
print("  trades :", nb_ia)

# ---- courbes ----
courbes = pd.DataFrame({
    "Buy and hold": (1 + rend_bh).cumprod(),
    "SMA200": (1 + rend_sma).cumprod(),
    "IA": (1 + rend_ia).cumprod(),
})
courbes.plot(logy=True, figsize=(10, 5))
plt.title("SPY : IA vs SMA200 vs buy and hold (walk-forward)")
plt.grid(True, alpha=0.3)
plt.show()
```
